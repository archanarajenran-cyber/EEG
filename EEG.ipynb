```python
import pandas as pd
import glob
import os

# Find all Advertisement 1 CSV files
ad1_files = glob.glob(
    r"C:\Users\admin\Downloads\Ads 1\Ads 1\*.csv"
)

# Create an empty list
ad1_list = []

# Load every CSV file
for file in ad1_files:
    data = pd.read_csv(file)
    ad1_list.append(data)

# Combine all files
ad1 = pd.concat(ad1_list, ignore_index=True)

print("Advertisement 1 loaded successfully!")
print("CSV files:", len(ad1_files))
print("Dataset Shape:", ad1.shape)
```

```python
print("Advertisement 1 Missing Values:")
print(ad1.isnull().sum().sum())

print("\nAdvertisement 1 Duplicate Rows:")
print(ad1.duplicated().sum())
```

```python
# Add advertisement labels
ad1['Advertisement'] = 'Advertisement 1'
ad2['Advertisement'] = 'Advertisement 2'

# Combine both datasets
combined_df = pd.concat([ad1, ad2], ignore_index=True)

# Display the result
print("Combined Dataset Shape:", combined_df.shape)

print("\nAdvertisement Counts:")
print(combined_df['Advertisement'].value_counts())
```

```python
print("Advertisement 1 Types:")
print(ad1['type'].value_counts())

print("\nAdvertisement 2 Types:")
print(ad2['type'].value_counts())
```

```python
brainwave_cols = ['alpha', 'beta', 'gamma', 'delta', 'theta']

print("Advertisement 1 - Average Brainwaves:")
print(ad1[brainwave_cols].mean())

print("\nAdvertisement 2 - Average Brainwaves:")
print(ad2[brainwave_cols].mean())
```

```python

other_cols = ['attention', 'stress', 'meditation',
              'blinkCount', 'fatigueRatio']

print("Advertisement 1:")
print(ad1[other_cols].mean())

print("\nAdvertisement 2:")
print(ad2[other_cols].mean())
```

```python
import matplotlib.pyplot as plt

attention_values = [
    ad1['attention'].mean(),
    ad2['attention'].mean()
]

plt.bar(['Advertisement 1', 'Advertisement 2'], attention_values)

plt.title('Average Attention Comparison')
plt.xlabel('Advertisements')
plt.ylabel('Average Attention')
plt.savefig('graph1_attention.png')

plt.show()
```

```python
stress_values = [
    ad1['stress'].mean(),
    ad2['stress'].mean()
]

plt.bar(['Advertisement 1', 'Advertisement 2'], stress_values)

plt.title('Average Stress Comparison')
plt.xlabel('Advertisements')
plt.ylabel('Average Stress')
plt.savefig('graph2_stress.png')

plt.show()
```

```python
brainwave_means = combined_df.groupby('Advertisement')[
    ['alpha', 'beta', 'gamma', 'delta', 'theta']
].mean()

brainwave_means.plot(kind='bar', figsize=(9, 5))

plt.title('Average Brainwave Comparison')
plt.xlabel('Advertisements')
plt.ylabel('Average Brainwave Values')
plt.xticks(rotation=0)
plt.legend(title='Brainwaves')
plt.tight_layout()
plt.savefig('graph3_brainwaves.png')
plt.show()
```

```python
meditation_values = [
    ad1['meditation'].mean(),
    ad2['meditation'].mean()
]

plt.bar(['Advertisement 1', 'Advertisement 2'], meditation_values)

plt.title('Average Meditation Comparison')
plt.xlabel('Advertisements')
plt.ylabel('Average Meditation')
plt.savefig('graph4_meditation.png')

plt.show()
```

```python
fatigue_values = [
    ad1['fatigueRatio'].mean(),
    ad2['fatigueRatio'].mean()
]

plt.bar(['Advertisement 1', 'Advertisement 2'], fatigue_values)

plt.title('Average Fatigue Ratio Comparison')
plt.xlabel('Advertisements')
plt.ylabel('Average Fatigue Ratio')
plt.savefig('graph5_fatigue.png')

plt.show()
```
